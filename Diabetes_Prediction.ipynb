```python
from sklearn.datasets import load_diabetes
```

```python
df = load_diabetes()
```

```python
print(df['DESCR'])
```

```python
df
```

```python
import pandas as pd
df_diabetes = pd.DataFrame(df.data, columns=['age' , 'sex' , 'bmi' , 'bp' , 's1' , 's2' , 's3' , 's4' , 's5' , 's6'])

df_diabetes.head()
df_diabetes.info()
```

```python
x = df_diabetes
y = df['target']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train.head()
```

```python
x_train.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))

sns.heatmap(x_train.corr() , annot = True)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor.fit(x_train, y_train)
```

```python
### Hyperparameter Tuning

param = {
    'criterion' : ['squared_error' , 'friedman_mse' , 'absolute_error'],
    'splitter' : ['best' , 'random'],
    'max_depth' : [1,2,3,4,5,10,15,20,25],
    'max_features': ['auto' , 'sqrt' , 'log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
regressor = DecisionTreeRegressor()

grid = GridSearchCV(regressor , param_grid = param , scoring = 'neg_mean_squared_error' , cv = 5 )
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(x_train , y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(x_test)
```

```python
from sklearn.metrics import r2_score , mean_absolute_error , mean_squared_error
print(r2_score(y_test , y_pred))
print(mean_absolute_error(y_test , y_pred))
print(mean_squared_error(y_test , y_pred))
```

```python
selected_model = DecisionTreeRegressor(criterion= 'squared_error',
 max_depth =  4,
 max_features =  'log2',
 splitter= 'best')
```

```python
selected_model.fit(x_train , y_train)
```

```python
import matplotlib.pyplot
from sklearn import tree

plt.figure(figsize = (15,10))
tree.plot_tree(selected_model , filled = True)
```

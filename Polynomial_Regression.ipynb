```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
x = 6*np.random.rand(100,1) - 3
y = 0.5*x**2 + 1.5*x + 2 + np.random.randn(100,1)

#Quadratic Eqaution used -:  y = 0.5x^2+1.5x+2+outliers
```

```python
plt.scatter(x,y,color ='g')
plt.xlabel('X Dataset')
plt.ylabel('Y Dataset')
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test , y_train , y_test = train_test_split(x,y,test_size = 0.3 , random_state = 42)

x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
regression1  = LinearRegression()
```

```python
regression1.fit(x_train , y_train)
```

```python
from sklearn.metrics import r2_score
sscore = r2_score(y_test , regression1.predict(x_test))
print(sscore)
```

```python
##Visualization of the Model

plt.plot(x_train , regression1.predict(x_train) , color = 'r')
plt.scatter(x_train , y_train)
```

```python
## Lets Apply Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree = 2 , include_bias = True)   ## Degree  = 2 means h_theta_x = h_0 + h1x1 + h2x^2
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.fit_transform(x_test)
```

```python
x_test_poly  ## [Bias , x1_Feature , x^2 Feature ]
```

```python
from sklearn.metrics import r2_score
regression = LinearRegression()
regression.fit(x_train_poly , y_train)
y_pred = regression.predict(x_test_poly)
score = r2_score(y_test , y_pred)
print(score)
```

```python
print(regression.coef_)
print(regression.intercept_)
```

```python
plt.scatter(x_train , regression.predict(x_train_poly))
plt.plot(x_train , y_train )
```
